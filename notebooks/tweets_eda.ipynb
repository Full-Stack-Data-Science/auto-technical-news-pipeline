```python
import pandas as pd
```

```python
file1 = "twitter_scraping_data_20260109_071513.parquet"
file2 = "twitter_scraping_data_20260109_191512.parquet"
```

```python
df_1 = pd.read_parquet(file1)
df_2 = pd.read_parquet(file2)
```

```python
df = pd.concat([df_2, df_1], ignore_index=True)
```

```python
df[df.author == 'karpathy']
```

```python
df = df.drop_duplicates(subset=['post_url', 'author'], keep='first').copy()
```

```python
df[df.author == 'karpathy']
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')

engagement_cols = ['replies', 'retweets', 'likes', 'bookmarks', 'views']
for col in engagement_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
TARGET_DAY = '2026-01-09'  # Change this to your desired date
target_dt = pd.Timestamp(TARGET_DAY).date()
filtered_df = df[df['date'].dt.date == target_dt].copy()
```

```python
filtered_df['total_engagement'] = filtered_df['views'] + filtered_df['replies'] + filtered_df['retweets'] + filtered_df['likes'] + filtered_df['bookmarks']
```

```python
filtered_df = filtered_df[filtered_df["is_tech_related"] == True]
```

```python
top_2_posts = filtered_df.nlargest(2, 'total_engagement')
```

```python
from openai import OpenAI

OPENAI_API_KEY="..."
def get_openai_client():
    return OpenAI(api_key=OPENAI_API_KEY)

client = get_openai_client()
```

```python
def openai_chat_complete(messages=(), model="gpt-4o-mini", raw=False):
    response = client.chat.completions.create(
        model=model,
        messages=messages
    )
    if raw:
        return response.choices[0].message
    output = response.choices[0].message
    return output.content
```

```python
def summarize_X_post():
    content_requirements = """
        CONTENT REQUIREMENTS:
        - Write in natural, fluent Vietnamese for a Vietnamese tech audience
        - Friendly, engaging tone
        - Total length: 200–300 words
        - Clearly explain why each post is trending, based on engagement metrics
        - Open with a short greeting to the tech community and mention that these are the hottest posts of the day
        """

    structure_requirements = """
        STRUCTURE REQUIREMENTS (MANDATORY):
        1. First paragraph:
           - A global headline for the entire summary
           - Wrapped in <p><strong>...</strong></p>

        2. For EACH post (Post 1 and Post 2), output in this exact order:
           - A post-specific title in <p><strong>Post X: TITLE</strong></p>
           - One or more <p> paragraphs summarizing the post
           - One <p> containing the post URL, written as a short call-to-action
             (example: Đọc thêm tại: https://...)

        3. Titles must be descriptive and meaningful, not generic labels.
        """

    format_requirements = """
        FORMATTING REQUIREMENTS (VERY IMPORTANT):
        - Output VALID HTML ONLY
        - Use <p> for all paragraphs
        - Use <strong> for titles and emphasis
        - Do NOT use Markdown syntax (###, **, *, _, -)
        - Do NOT use unsupported HTML tags
        - Do NOT include code blocks or backticks
        """

    user_prompt = f"""You are a tech content curator.
        Create an engaging summary for a tech community channel based on the top 2 posts from {TARGET_DAY}.

        Post 1:
        {top_2_posts.iloc[0].to_json()}

        Post 2:
        {top_2_posts.iloc[1].to_json()}

        {content_requirements}
        {structure_requirements}
        {format_requirements}

        The output must be ready to paste directly into a rich-text editor.
        """

    openai_messages = [
        {
            "role": "system",
            "content": "You write clean, well-structured HTML content for Vietnamese tech news communities."
        },
        {"role": "user", "content": user_prompt}
    ]

    return openai_chat_complete(openai_messages)
```
